```python
df_drop = ['ADDRESSLINE1', 'ADDRESSLINE2', 'STATUS', 'POSTALCODE',
           'CITY', 'TERRITORY', 'PHONE', 'STATE',
           'CONTACTFIRSTNAME', 'CONTACTLASTNAME',
           'CUSTOMERNAME', 'ORDERNUMBER', 'PRODUCTCODE']

df = df.drop(df_drop, axis=1)
```

```python
productline = pd.get_dummies(df['PRODUCTLINE'])
Dealsize = pd.get_dummies(df['DEALSIZE'])

# Converting categorical columns.
```

```python
df = pd.concat([df, productline, Dealsize], axis=1)
```

```python
df_drop = ['COUNTRY', 'PRODUCTLINE', 'DEALSIZE']

df = df.drop(df_drop, axis=1) #Dropping Country too as there are alot of countries. 
```

```python
df.drop('ORDERDATE', axis=1, inplace=True)
# Dropping ORDERDATE as month, quarter and year are already included.
```

```python
df.dtypes #All the datatypes are converted into numeric
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(df)

# Scaling the features so that all features have equal importance.
```

```python
distortions = [] # Within Cluster Sum of Squares from the centroid

K = range(1, 10)

for k in K:
    kmeanModel = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeanModel.fit(X_train)
    distortions.append(kmeanModel.inertia_)
```

```python
plt.figure(figsize=(16,8))
plt.plot(K, distortions, 'bx-')
plt.xlabel('k')
plt.ylabel('Distortion')
plt.title('The Elbow Method showing the optimal k')
plt.show()
```

```python
X_train.shape
```

```python
model = KMeans(n_clusters=3, random_state=42, n_init=10)

model = model.fit(X_train)

predictions = model.predict(X_train)

# Predicting the cluster values (0, 1, or 2).
```

```python
unique, counts = np.unique(predictions, return_counts=True)
```

```python
counts = counts.reshape(1, 3)
```

```python
counts_df = pd.DataFrame(
    counts,
    columns=['Cluster1', 'Cluster2', 'Cluster3']
)
```

```python
counts_df.head()
```

```python
pca = PCA(n_components=2)

# Converting all features into two columns for visualization.
```

```python
reduced_X = pd.DataFrame(
    pca.fit_transform(X_train),
    columns=['PCA1', 'PCA2']
)
```

```python
reduced_X.head()
```

```python
# Plotting the normal scatter plot.

plt.figure(figsize=(14,10))
plt.scatter(reduced_X['PCA1'], reduced_X['PCA2'])
plt.xlabel('PCA1')
plt.ylabel('PCA2')
plt.title('PCA Visualization')
plt.show()
```

```python
model.cluster_centers_
```

```python
reduced_centers = pca.transform(model.cluster_centers_)
```

```python
reduced_centers
```

```python
plt.figure(figsize=(14,10))

plt.scatter(reduced_X['PCA1'], reduced_X['PCA2'])

plt.scatter(
    reduced_centers[:,0],
    reduced_centers[:,1],
    color='black',
    marker='x',
    s=300
)

plt.title('K-Means Cluster Centroids')
plt.show()
```

```python
reduced_X['Clusters'] = predictions

reduced_X.head()
```

```python
# Plotting the clusters.

plt.figure(figsize=(14,10))

plt.scatter(
    reduced_X[reduced_X['Clusters'] == 0]['PCA1'],
    reduced_X[reduced_X['Clusters'] == 0]['PCA2'],
    color='slateblue'
)

plt.scatter(
    reduced_X[reduced_X['Clusters'] == 1]['PCA1'],
    reduced_X[reduced_X['Clusters'] == 1]['PCA2'],
    color='springgreen'
)

plt.scatter(
    reduced_X[reduced_X['Clusters'] == 2]['PCA1'],
    reduced_X[reduced_X['Clusters'] == 2]['PCA2'],
    color='indigo'
)

plt.scatter(
    reduced_centers[:,0],
    reduced_centers[:,1],
    color='black',
    marker='x',
    s=300
)

plt.xlabel('PCA1')
plt.ylabel('PCA2')
plt.title('K-Means Clustering Visualization')
plt.show()
```

```python
#Hierarchical Clustering
#Hierarchical clustering groups similar data points into clusters by building a hierarchy. 
#We use the Ward linkage method to minimize the variance within clusters.
Z = linkage(X_train, method='ward')

plt.figure(figsize=(16,8))

dendrogram(
    Z,
    truncate_mode='lastp',
    p=30
)

plt.title('Dendrogram for Hierarchical Clustering')
plt.xlabel('Data Points')
plt.ylabel('Euclidean Distance')
plt.show()
```

```python
#Applying Agglomerative Clustering
hc = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

hc_predictions = hc.fit_predict(X_train)
```

```python
#Finding the number of data points in each cluster
unique, counts = np.unique(
    hc_predictions,
    return_counts=True
)
```

```python
counts = counts.reshape(1, 3)
```

```python
counts_df = pd.DataFrame(
    counts,
    columns=['Cluster1', 'Cluster2', 'Cluster3']
)

counts_df.head()
```

```python
#Visualizing Hierarchical Clusters using PCA
reduced_X['Hierarchical_Clusters'] = hc_predictions

reduced_X.head()
```

```python
plt.figure(figsize=(14,10))

plt.scatter(
    reduced_X[reduced_X['Hierarchical_Clusters'] == 0]['PCA1'],
    reduced_X[reduced_X['Hierarchical_Clusters'] == 0]['PCA2'],
    color='slateblue'
)

plt.scatter(
    reduced_X[reduced_X['Hierarchical_Clusters'] == 1]['PCA1'],
    reduced_X[reduced_X['Hierarchical_Clusters'] == 1]['PCA2'],
    color='springgreen'
)

plt.scatter(
    reduced_X[reduced_X['Hierarchical_Clusters'] == 2]['PCA1'],
    reduced_X[reduced_X['Hierarchical_Clusters'] == 2]['PCA2'],
    color='indigo'
)

plt.xlabel('PCA1')
plt.ylabel('PCA2')
plt.title('Hierarchical Clustering Visualization')
plt.show()
```
